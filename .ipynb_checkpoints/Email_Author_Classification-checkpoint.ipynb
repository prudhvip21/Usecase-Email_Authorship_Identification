```python
import os
os.chdir('maildir')
names= os.listdir()
```

```python
file_name= list()
content= list()
for name in names:
  if '_sent_mail' in os.listdir(f'{name}'):
    for files in os.listdir(f'{name}/_sent_mail'):
      f= open(f'{name}/_sent_mail/{files}', 'r', encoding= 'utf-8')
      file_name.append(f'{name}/_sent_mail/{files}')
      content.append(f.read())
```

```python
df= pd.DataFrame({'file_path': file_name, 'content': content})
```

# Data Description

This dataset was collected and prepared by the <a href="http://www.ai.sri.com/project/CALO">CALO Project</a> (A
    Cognitive Assistant that Learns and Organizes).  It contains data
    from about 150 users, mostly senior management of Enron, 
    organized into folders.  The corpus contains 
    a total of about 0.5M messages.
    This data was originally <a href="http://www.salon.com/news/feature/2003/10/14/enron/index_np.html">made
    public, and posted to the web</a>, by the <a href="http://www.ferc.gov">Federal Energy Regulatory
    Commission</a> during its investigation.

  <p>
    The email dataset was later purchased by <a href="http://www.ai.mit.edu/people/lpk/lpk.html">Leslie
    Kaelbling</a> at MIT, and turned out to have a number of integrity
    problems.  A number of folks at SRI, notably <a href="http://www.ai.sri.com/people/gervasio">Melinda Gervasio</a>,
    worked hard to correct these problems, and it is thanks to them
    (not me) that the dataset is available.  The dataset here does not
    include attachments, and some messages have been deleted "as part
    of a redaction effort due to requests from affected
    employees". Invalid email addresses were converted to something of
    the form user@enron.com whenever possible (i.e., recipient is
    specified in some parse-able format like "Doe, John" or "Mary
    K. Smith") and to no_address@enron.com when no recipient was
    specified.</p>

> Credits: [Enron](https://www.cs.cmu.edu/~enron/)

## Extracting ID

```python
def find_ID(text):
  pattern= r'Message-ID: \<(.*?)\>'
  text= re.findall(pattern, text)
  return text[0]
df['id']= df.content.apply(find_ID)
```

## Extracting Date

```python
def find_date(text):
  pattern= r'Date: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['date']= df.content.apply(find_date)
```

## Extracting Sent From Email-Id

```python
def email_from(text):
  pattern= r'From: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['from_email']= df.content.apply(email_from)
```

## Extracting Sent To Email-Id

```python
def email_to(text):
  pattern= r'To: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['to_email']= df.content.apply(email_to)
```

## Subject

```python
def ex_sub(text):
  pattern= r'Subject: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['subject']= df.content.apply(ex_sub)
```

## Is reply

```python
df['is_reply']= df.subject.str.contains('Re:')
```

```python
df['is_reply']= df.is_reply.map({True: 1, False: 0})
```

## Email From Name

```python
def name_from(text):
  pattern= r'X-From: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['from_name']= df.content.apply(name_from)
```

## Email To Name

```python
def name_to(text):
  pattern= r'X-To: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['to_name']= df.content.apply(name_to)
```

## Extracting Content

```python
df['message']= df.content.str.split('X-FileName:')
df['message']= df.message.apply(lambda x: x[1])
df['message']= df.message.str.split('\n\n', n= 1)
df['message']= df.message.apply(lambda x: x[1])
```
