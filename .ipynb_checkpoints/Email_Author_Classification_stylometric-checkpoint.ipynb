It is a very quick, asynchronous written communication channel that is used for various purposes  ranging from formal to informal communication.*

```python
def extract_author(text):
  pattern= r'maildir/(.*?)/'
  text= re.findall(pattern, text)
  return text[0]
```

## Extracting ID

```python
def find_ID(text):
  pattern= r'Message-ID: \<(.*?)\>'
  text= re.findall(pattern, text)
  return text[0]
df['id']= df.content.apply(find_ID)
```

*This pattern/regex is searching for all the characters between `Message-ID: <` and `>`, backshlashes before `<` and `>` is to escpae it.* 

## Extracting Date

```python
def find_date(text):
  pattern= r'Date: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['date']= df.content.apply(find_date)
```

*This above regex is also extracting characters between `Date: ` and `\n`*

*Rest all the features we will be extracting follows the same pattern.*

## Extracting Sent From Email-Id

```python
def email_from(text):
  pattern= r'From: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['from_email']= df.content.apply(email_from)
```

## Extracting Sent To Email-Id

```python
def email_to(text):
  pattern= r'To: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['to_email']= df.content.apply(email_to)
```

## Subject

```python
def ex_sub(text):
  pattern= r'Subject: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['subject']= df.content.apply(ex_sub)
```

## Is reply

```python
df['is_reply']= df.subject.str.contains('Re:')
```

*This checks wheather your email is a reply. And then if it is a reply map it to 1 and if not then map it to zero.*

```python
df['is_reply']= df.is_reply.map({True: 1, False: 0})
```

## Email From Name

```python
def name_from(text):
  pattern= r'X-From: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['from_name']= df.content.apply(name_from)
```

## Email To Name

```python
def name_to(text):
  pattern= r'X-To: (.*?)\n'
  text= re.findall(pattern, text)
  return text[0]
df['to_name']= df.content.apply(name_to)
```

### Creating Stylometric Features

**Email Length**

```python
data['email_length'] = data.apply(lambda x : len(x['message']) , axis = 1)
```

**Digits Count**

```python
data['digits_count'] = data.apply(lambda x : sum(c.isdigit() for c in x['message']) , axis = 1)
```

```python
data['digits_count']  = data['digits_count'] / data['email_length']
```

```python
sns.set(rc={'figure.figsize':(19,7)})
sns.set_style('white')
sns.violinplot(x="author", y="digits_count", data=data)
plt.title("Normalised Digits Count Distribution", fontsize= 18)
plt.xlabel("Authors", fontsize= 15)
plt.ylabel("Digits Count", fontsize= 15)
plt.show()
#put y limit = 0.5 and see
```

**Spaces Count**

```python
data['spaces_count'] = data.apply(lambda x : sum(c.isspace() for c in x['message']) , axis = 1)
```

```python
data['spaces_count']  = data['spaces_count'] /data['email_length']
```

```python
sns.set(rc={'figure.figsize':(19,7)})
sns.set_style('white')
sns.violinplot(x="author", y="spaces_count", data=data)
plt.title("Normalised Spaces counts Distribution", fontsize= 18)
plt.xlabel("Authors", fontsize= 15)
plt.ylabel("Spaces count", fontsize= 15)
plt.show()
```

**Words Count**

```python
data['word_count'] = data.apply(lambda x : len(re.findall(r'\w+', x['message'])) , axis = 1)
```

```python
data['word_count'] = data['word_count'] /data['email_length']
```

```python
sns.set(rc={'figure.figsize':(19,7)})
sns.set_style('white')
sns.violinplot(x="author", y="word_count", data=data)
plt.title("Normalised Word Counts Distribution", fontsize= 18)
plt.xlabel("Authors", fontsize= 15)
plt.ylabel("Word Count", fontsize= 15)
plt.show()
```

**Average Word Length**

```python
def avg_word_length(row):
    filtered_sentence = ''.join([x for x in row['message'] if x not in '".,;!-'])
    words = filtered_sentence.split()
    try :
        avg = sum(map(len, words)) / len(words)
    except :
        avg = 0
    return avg
```

```python
data['avg_word_length'] = data.apply(lambda x : avg_word_length(x) ,axis = 1)
```

```python
sns.set(rc={'figure.figsize':(19,7)})
sns.set_style('white')
sns.violinplot(x="author", y="avg_word_length", data=data)
plt.title("Average Word Length", fontsize= 18)
plt.xlabel("Authors", fontsize= 15)
plt.ylabel('')
plt.show()
```

****

```python
def avg_sent_length(row) :
    sents = sent_tokenize(row['message'])
    lengths_of_sents = [len(k) for k in sents]
    avg_length = sum(lengths_of_sents) / len(sents)
    return  avg_length
```

```python
data['avg_sentence_length'] = data.apply(lambda x : avg_sent_length(x) ,axis = 1)
```

```python
sns.set(rc={'figure.figsize':(19,7)})
sns.set_style('white')
sns.violinplot(x="author", y="avg_sentence_length", data=data)
plt.title("Average Sentence Length", fontsize= 18)
plt.xlabel("Authors", fontsize= 15)
plt.ylabel('')
plt.show()
```

**Special Character Counts**

```python
special_chars = string.punctuation
data['spl_char_count'] = data.apply(lambda  x : sum(v for k, v in list(Counter(x['message']).items()) if k in special_chars) , axis = 1)
data['spl_char_count'] = data['spl_char_count'] / data['email_length']
```

```python
sns.set(rc={'figure.figsize':(19,7)})
sns.set_style('white')
sns.violinplot(x="author", y="spl_char_count", data=data)
plt.title("Normalised Special Character Counts", fontsize= 18)
plt.xlabel("Authors", fontsize= 15)
plt.ylabel('Special Chracter Counts')
plt.show()
```

**Average number of sentences in Paragraph.**

```python
data['paras_count'] = data.apply(lambda x : x['message'].count('\n\n') + 1 ,axis = 1)
```

```python
data['avg_sentences_in_para'] = data.apply(lambda x :len(re.findall(r'\.',x['message']))/x['paras_count'],axis = 1)
```

```python
sns.set(rc={'figure.figsize':(19,7)})
sns.set_style('white')
sns.violinplot(x="author", y="avg_sentences_in_para", data=data)
plt.title("Average Sentences in a Paragraph", fontsize= 18)
plt.xlabel("Authors", fontsize= 15)
plt.ylabel('Sentences in a para')
plt.show()
```

**Ratio of Stopwords**

```python
stops = stopwords.words('english')
def ratio_fun_words(row) :
    count = len([i for i in word_tokenize(row['message'].lower()) if i in stops])
    try :
        count = count/(row['word_count']*row['email_length'])
    except :
        count = 0

    return  count
```

```python
data['ratio_of_fun_words'] = data.apply(lambda x : ratio_fun_words(x),axis = 1)
```

```python
sns.set(rc={'figure.figsize':(19,7)})
sns.set_style('white')
sns.violinplot(x="author", y="ratio_of_fun_words", data=data)
plt.title("Ratio of Stop Words", fontsize= 18)
plt.xlabel("Authors", fontsize= 15)
plt.ylabel('')
plt.show()
```

## Text Cleaning

*This Clean Message function is first substituting everything that is not a chracter with space and then removing the stopwords.*

```python
def clean_message(content, remove_stopwords= True):
  content = re.sub("[^a-zA-Z]"," ", content)
  words = word_tokenize(content.lower())
    
  if remove_stopwords:
      stops = set(stopwords.words("english"))
      words = [w for w in words if not w in stops]
  return ' '.join(words)
```

```python
data.message= data.message.apply(clean_message)
```

## Building Classification Models

### Logistic Regression

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42, stratify= y)
```

```python
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state=42)
classifier.fit(X_train, y_train)
```

```python
y_pred= classifier.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(y_pred, y_test))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred, target_names= list(le.classes_)))
```

Some algorithms (such as Random Forest classifiers or naive Bayes classifiers) are capable of handling multiple classes directly. Others (such as Support Vector Machine classifiers or Linear classifiers) are strictly binary classifiers.

*So now the question is How binary classifiers likes Logistic Regression and Support Vector Machines can classify multiple classes.*

There are two approaches to deal with this 

- `OneVsOneClassifier`
- `OneVsRestClassifier`

**OneVsRestClassifier**

*one way to create a system that can classify the Authors Names into 10 classes (from 0 to 9) is to train 10 binary classifiers, one for each Author (a 0-detector, a 1-detector, a 2-detector, and so on). Then when you want to classify an image, you get the decision score from each classifier for that image and you select the class whose classifier outputs the highest score.*

**OneVsOneClassifier**

*Another strategy is to train a binary classifier for every pair of Authors: one to distinguish 0s and 1s, another to distinguish 0s and 2s, another for 1s and 2s, and so on. This is called the one-versus-one (OvO) strategy.So, If there are N classes, you need to train `N × (N – 1) / 2` classifiers.*


**Scikit-Learn detects when you try to use a binary classifier algorithm for a multi-class classification task, and it automatically runs OneVsRestClassifier or OneVsOneClassifier, depending on the algorithm.**



*Now we have a problem here for some authors the classifier is working perfect but for other the classifier is under performing. So accuracy won't be great measure.*

So Now Another question When to use Micro and when to use Micro!!

`Macro-average method can be used when you want to know how the system performs overall across the sets of data.`

`On the other hand, micro-average can be a useful measure when your dataset varies in size.`

**We will use macro average F1 Score as our evaluation metrics.**

```python
from sklearn.metrics import f1_score
print(f1_score(y_pred, y_test, average= 'macro'))
```
